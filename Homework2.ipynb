```python
## Preparing dataset
import pandas as pd
import numpy as np
df = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")
df = df[['engine_displacement','horsepower','vehicle_weight','model_year','fuel_efficiency_mpg']]
```

```python
# Q1
print(df.count())
```

```python
# Q2
hp_q50 = df['horsepower'].median() 
hp_q50
```

```python
# Prepare for Q3 to 6
n = len(df)
n_val = int(n * 0.2)
n_test = int(n * 0.2)
n_train = n - n_val - n_test

np.random.seed(42)
idx = np.arange(n)
np.random.shuffle(idx)

df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train + n_val]]
df_test = df.iloc[idx[n_train + n_val:]]
```

```python
# Fill horsepower with 0
# create new df's to not overwrite existing
df_train0 = df_train
df_val0 = df_val
df_test0 = df_test

df_train0['horsepower'] = df_train0['horsepower'].fillna(0)
df_val0['horsepower'] = df_val0['horsepower'].fillna(0)
df_val0['horsepower'] = df_test0['horsepower'].fillna(0)



```

```python
# Fill horsepower with mean
df_train_q50 = df_train
df_val_q50 = df_val
df_test_q50 = df_test

df_train_q50['horsepower'] = df_train_q50['horsepower'].fillna(hp_q50)
df_val_q50['horsepower'] = df_val_q50['horsepower'].fillna(hp_q50)
df_test_q50['horsepower'] = df_test_q50['horsepower'].fillna(hp_q50)
```
